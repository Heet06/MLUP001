# Data Preparation

```python
import pandas as pd
import numpy as np
```

### Load data

```python
demoDetails = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/datasets/demoDetails.csv", index_col=0)
acDetails = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/datasets/acDetails.txt", sep="\t")
serviceDetails = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/datasets/serviceDetails.csv", index_col=0)
```

By setting `index_col=0`, 1st column will be the index column

### Data Wrangling

- We are interested in merging acDetails, demoDetails and serviceDetails
- Before merging we need to make necesarry checks!
- What are mandatory checks before merging?
  1. Are there any duplicate records?
  2. Whether the customer ID common across all the files?

#### **1. Are there any duplicate records?**

```python
# Are there any duplicate records?
print(len(demoDetails))
len(np.unique(demoDetails['customerID']))
```

```python
print(len(acDetails))
len(np.unique(acDetails['customerID']))
```

```python
print(len(serviceDetails))
len(np.unique(serviceDetails['customerID']))
```

Yes, there is one duplicate record across all the three dataframes

```python
# To determine the duplicate records 'duplicated()' method can be used

demoDetails.duplicated(subset=['customerID'], keep=False)

# This method returns a boolean series with True value for each duplicated row
```

```python
# So now let's subset the rows and look at the duplications
```

```python
demoDetails[demoDetails.duplicated(['customerID'], keep=False)]
```

```python
acDetails[acDetails.duplicated(['customerID'], keep=False)]
```

```python
acDetails[acDetails.duplicated(['customerID'], keep=False)]
```

```python
# Removing Duplicate Records

demoDetails = demoDetails.drop_duplicates()

acDetails = acDetails.drop_duplicates()

serviceDetails = serviceDetails.drop_duplicates()
```

By default the first occurence of duplicate rows is kept and subsequent occurence are removed.

`keep` parameter can be used to:
- 'last': keep the last occurence
- 'first': keep the first occurence
- 'False': all occurences other than last one are removed

#### **2. Whether customerID is common across all the files?**

syntax: `dataframe1.equals(dataframe2)`

```python
acDetails.customerID.equals(demoDetails.customerID)
```

```python
serviceDetails.customerID.equals(demoDetails.customerID)
```

```python
acDetails.customerID.equals(serviceDetails.customerID)
```

customerID in all files/datasets are identical.

### Joining three dataframes

Syntax: `pd.merge(df1, df2, on=[ColumnName], how='inner')`

```python
churn = pd.merge(demoDetails, acDetails, on = 'customerID')

churn = pd.merge(churn, serviceDetails, on = 'customerID')

churn1 = churn.copy()

churn.columns
```
