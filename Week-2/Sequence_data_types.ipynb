2. String Operations

```python
language = "Python"
upper_text = language.upper()
print(upper_text)
```

```python
lower_text = upper_text.lower()
print(lower_text)
```

```python
print(lower_text.replace('p', 'P'))
```

```python
print("length of lower text is: ", len(lower_text))
```

```python
h = print("GM")
```

```python
print(h)
```

"None" - In Python, the "None" value doesn't have a length because it represents the absence of a value or a null value. It is not a sequence or container type that can be measured in terms of length.

If you try to get the length of None using the len() function, you will encounter a TypeError because None does not suppot the len() operation. Here's an example.

```python
x = None
print(len(x))
```

3. String Indexing and Slicing

```python
first_letter, last_letter = language[0], language[-1]
print(first_letter, last_letter)
```

```python
first_four_letters = upper_text[:4]
last_four_letters = upper_text[-2::-1]
print(first_four_letters, last_four_letters)
```

4. String Concatenation

```python
language = "Python"
print(language, id(language))
language += " Programming"
print(language, id(language))
```

- ** Note: The id() method returns the identity  of the object (address of an object in memory)**
- This proves strings are immutable

**Syntax:** object * integer

```python
language = "Python"
language *= 3
```

```python
print(language)
```

5. String Slicing
  - : is used to slice some element from an ordered sequence
    - Syntax: object[start:stop:step]
    - The default value for start and step are 0 and 1 respectively
    - Negative value for step can be used to access the elements from the end of the sequence

```python
samplestr = 'Python'
samplestr[slice(4)]
```

```python
samplestr[slice(1,6,2)]
```
