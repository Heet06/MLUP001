# Introduction to NumPy

Numpy is a foundation of numerical computing in Python. It powers data analysis, machine learning and deep learning workflows.

#### Import Numpy

```python
import numpy as np
print("NumPy version:", np.__version__)
```

What is an "array"?

In programming, an array is a way to store and organize data — think of it like a list or table. In NumPy, arrays can have any number of dimensions:

1D: a list of values

2D: like a table (rows and columns)

3D: a stack of tables (like pages)

In NumPy, arrays are called ndarray (N-dimensional arrays)

Key rules: All items must be the same data type

The array has a fixed size

Shape must be regular (no jagged edges)

#### Creating arrays

1. From list:
  Create a numpy array from a Python list or from a sequence of elements

```python
my_list = [1, 2, 3, 4, 5, 6]
print("Python list:", my_list)

array = np.array(my_list)
print("NumPy array:", array)
```

```python
print(array[0])

array[0] = 10
print(array)

b = array[:3]
print(b)
```

**Creating Arrays from Nested Lists**

Simulating small tabular datasets.

```python
nested_list = [
    [1, 2, 3],
    [4, 5, 6]
]

multi_array = np.array(nested_list)
print("Multi-dimensional array:\n", multi_array)
```

In NumPy, "axis" = dimension (e.g., rows = axis 0, columns = axis 1)

Access elements in muti-dimensional arrays like: a[row,column]

```python
multi_array[1,2]
```

#### Array attributes

Key attributes of NumPy arrays:
  - `ndim`: number of dimensions
  - `shape`: dimensions along each axis
  - `size`: total number of elements in array
  - `dtype`: data type of elements

```python
print("Type:", type(multi_array))
print("Number of dimensions:", multi_array.ndim)
print("Shape:", multi_array.shape)
print("Size:", multi_array.size)
print("Data Type:", multi_array.dtype)
```

#### Creating Basic Arrays

1. Create an array filled with 0's

```python
zeros_1 = np.zeros(3)
zeros_2 = np.zeros((3, 3))
print(zeros_1)
print(zeros_2)
```

2. Creating an array filled with 1's:

```python
ones_1 = np.ones(2)
ones_2 = np.ones((2, 4))
print(ones_1)
print(ones_2)
```

3. Creating an empty array:

```python
np.empty(2)
```

4. Create an array with a range of elements

```python
range_arr = np.arange(4)
range_arr
```

5. Create an array with evenly spaced values:

Use np.arange(start, stop, step)

Create an array from 0 to 18 with step size 3.

```python
range_arr =np.arange(0, 21, 3)
print(range_arr)
```

6. Create an array with evenly spaced values

Use np.linspace(start, stop, num)

```python
np.linspace(0, 10, num=5)
```

**Specifying your data type:**

Use the dtype keyword in NumPy to set the data type of an array.

Default: floating point (np.float64)

```python
ones_1 = np.ones(2)
ones_1
```

```python
ones_1 = np.ones(2, dtype=np.int32)
ones_1
```

#### Adding, Removing and Sorting Elements

**Sort:**

np.sort(arr) sorts values in ascending order

```python
arr = np.array([2, 1, 5])
np.sort(arr)
```

**Join arrays:**

np.concatenate((a,b)) joins arrays along an axis

```python
x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6]])
np.concatenate((x, y), axis=0)
```
