### List

  - A sequence data type which can contain multiple items of different data types
  - A list can be created using a square bracket `[]` or using the `list()` method
  - In general, lists are ordered, mutable, and can take duplicate elements.

1. Intialization

```python
empty_list = []
numbers = [1, 2, 3, 4, 5]
mixed = [1, "apple", 3.14, True]
nested = [[1, 2], [3, 4], [5, 6]]
```

```python
print(f"Empty List: {empty_list}")
print(f"Numbers: {numbers}")
print(f"Mixed: {mixed}")
print(f"Nested: {nested}")
```

```python
string_value = "hello"
char_list = list(string_value)
print(char_list)
```

2. List Operations

```python
value = 7
numbers.append(value)
print(numbers)
```

```python
mixed.reverse()
print(mixed)
```

2. List Indexing and Slicing

```python
second_item = mixed[2]
third_item_from_last = mixed[-2]
print(second_item, third_item_from_last)
```

```python
first_three_items = mixed[:3]
last_three_items = mixed[-4:-1]
print(first_three_items, last_three_items)
```

```python
lstSample = [50, 100, 'C', 'team', 'India']
```

```python
lstSample.index('team')
```

```python
lstSample.index('India')
```

Inserting elements at a specific index using `insert()`:

```python
print(mixed)
```

```python
mixed.insert(2, 'New Zealand')
print(f"items = {mixed}")
```

```python
mixed.remove('Australia')
```

```python
mixed.pop(2)
print(f"After popping out 2nd element: items= {mixed}")
```

4. List Concatenation

```python
l1 = [1, 3, 5, 7]
print(l1, id(l1))
l1 += [2, 4, 6, 8]
print(l1, id(l1))
```

### Tuples
  - A sequence data type which contain multiple items of different data types
  - A tuple can be created using parenthesis `()` or using the `tuple()` method
  - In general, tuples are ordered, immutable and can take duplicate values

1. Initialization

```python
account_balance = (10302, 57059, 8030, 14688, 103023, 28734, 57059, 8030, 14688)
print(account_balance, type(account_balance))
print(len(account_balance))
```

```python
account_holders = ('India', ['Stella', 'Rohit', 'Jessi'])
print(account_holders, type(account_holders))
print(len(account_holders))
```

2. Tuple Operations

```python
print(account_balance.index(10302))
```

  - Unpacking tuples - You can assign the elements of a tuple to multiple variables in a single assignment statement

```python
account_details = ('Virat', 32, 1234)
name, age, balance = account_details

print(name)
print(age)
print(balance)
```

3. Tuple Indexing and Slicing

```python
first_account_balance = account_balance[0]
last_account_balance = account_balance[-1]
print(first_account_balance, last_account_balance)
```

```python
first_four_accounts_balance = account_balance[:5]
last_four_accounts_balance = account_balance[-4:]
print(first_four_accounts_balance, last_four_accounts_balance)
```

4. Tuple Concatenation

```python
a = (4)
type(a)
```

```python
tup = 1, 2, 3
print(tup, id(tup))
tup += (4,)
print(tup, id(tup))
```

  - This proves tuples are immutable

### Dictionary
  - A sequence data type which can contain multiple items of different data types. The items are stored as key:value pairs
  - A dictionary can also be created using flower bracket `{}` or by using the `dict()` method
  - In general, dictionaries are ordered, mutable, and can not take duplicate values. By order, it means that the items in dictionaries are retained in the same order as they were when created

1. Intialization

```python
car_details = {
    "Brand": "Hyundai",
    "Model": "i20",
    2: 821695
}
```

```python
car_details2 = dict([
    ("Brand", "Hyundai"),
    ("Model", "i20"),
    (2, 821695)
])
```

```python
print(car_details["Brand"])
print(car_details2[2])
```

2. Dictionary Operations

```python
car_details.get("Brand")
```

```python
print(car_details.keys())
print(car_details.values())
print(car_details.items())
```

```python
print(car_details.get('name'))
print(car_details.get('name', 0))
```

```python
car_details.items()
```

```python
for key, val in car_details.items():
  print(key, ": ", val)
```

```python
value = car_details.pop('Model')
print(value)

key, value = car_details.popitem()
print(key)
print(value)
print(car_details)
```

3. Dictionary Concatenation

```python
car_details.update({'Price':"1135682"})
print(car_details)
```

```python
car_details['milaege'] = 19
print(car_details['milaege'])
```

### Range
  - A range is a sequence of numbers that can be created using the `range()` function
  - Commonly used in loops and is especially helpful for iterating through a range of numbers
  - Function can take one, two or three arguments, and it returns a sequence of numbers starting from a specific start value up to (but not including) an end value, and with a specific step value
    - `range(stop)`
    - `range(start,stop)`
    - `range(start, stop, step)`

1. Initialization

```python
r1 = range(5)
r1
```

```python
for i in r1:
  print(i)
```

```python
r2 = range(2, 7)
r2
for i in r2:
  print(i)
```

```python
r3 = range(1, 12, 2)
print(r3)
for i in r3:
  print(i)
```
